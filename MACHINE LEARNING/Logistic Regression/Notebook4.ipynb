```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

```python
data = {
    "Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Pass":  [0, 0, 0, 0, 1, 1, 1, 1]
}

df = pd.DataFrame(data)
df
```

```python
X = df[["Hours"]]     
y = df["Pass"]        
```

```python
X
```

```python
y
```

```python
model = LogisticRegression()
```

```python
model.fit(X, y)
```

```python
print(model.coef_)
```

```python
print(model.intercept_)
```

```python
prediction = model.predict([[5]])
print(prediction)
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(1, 8, 100).reshape(-1, 1)

plt.scatter(X, y)
plt.plot(x, model.predict_proba(x)[:, 1], 'r')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate smooth values for the curve
x = np.linspace(0, 9, 100).reshape(-1, 1)

# Predict probabilities
y_prob = model.predict_proba(x)[:, 1]

# Plot data points
plt.scatter(X, y, color='blue', label='Actual Data')

# Plot sigmoid curve
plt.plot(x, y_prob, color='red', linewidth=2, label='Logistic Regression')

plt.xlabel("Study Hours")
plt.ylabel("Probability of Passing")
plt.title("Binary Logistic Regression")
plt.legend()
plt.grid(True)

plt.show()
```

```python
from mlxtend.plotting import plot_decision_regions
import matplotlib.pyplot as plt

plot_decision_regions(X.to_numpy(), y.to_numpy(), clf=model)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Decision Boundary")
plt.show()
```
