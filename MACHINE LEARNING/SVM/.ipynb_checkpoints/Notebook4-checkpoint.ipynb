### Hard Margin SVM

#### Very Large C (1e6)
- Misclassification is heavily penalized.
- The model tries to classify every training point correctly.
- The margin becomes as small as needed to avoid mistakes.
- This behavior closely approximates a Hard Margin SVM.

```python
from sklearn.datasets import make_classification
from sklearn.svm import SVC
import matplotlib.pyplot as plt
import numpy as np

# Perfectly separable dataset
X, y = make_classification(
    n_samples=100,
    n_features=2,
    n_redundant=0,
    n_clusters_per_class=1,
    class_sep=3,
    random_state=42
)

model = SVC(kernel='linear', C=1e6)
model.fit(X, y)

plt.scatter(X[:,0], X[:,1], c=y, cmap='coolwarm')
plt.title("Hard Margin SVM")
plt.show()
```

### Soft Margin SVM
Data contains noisy overlapping classes

```python
X, y = make_classification(
    n_samples=200,
    n_features=2,
    n_redundant=0,
    class_sep=0.8,
    flip_y=0.1,
    random_state=42
)

model = SVC(kernel='linear', C=1)
model.fit(X, y)

plt.scatter(X[:,0], X[:,1], c=y, cmap='coolwarm')
plt.title("Soft Margin SVM")
plt.show()
```

### Using Polynomial kernel

```python
from sklearn.datasets import make_circles

X, y = make_circles(
    n_samples=300,
    noise=0.08,
    factor=0.3,
    random_state=42
)

model = SVC(
    kernel='poly',
    degree=3
)

model.fit(X, y)

plt.scatter(X[:,0], X[:,1], c=y, cmap='coolwarm')
plt.title("Polynomial Kernel SVM")
plt.show()
```

```python
#Visualize support vectors

plt.scatter(X[:,0], X[:,1], c=y)

plt.scatter(
    model.support_vectors_[:,0],
    model.support_vectors_[:,1],
    s=180,
    facecolors='none',
    edgecolors='black',
    linewidth=2,
    label='Support Vectors'
)

plt.legend()
```
