```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_circles
from sklearn.decomposition import PCA, KernelPCA
```

```python
X, y = make_circles(
    n_samples=500,
    noise=0.05,
    factor=0.3,
    random_state=42
)
```

```python
plt.figure(figsize=(6,6))
plt.scatter(X[:,0], X[:,1], c=y, cmap='viridis')
plt.title("Original Non-linear Data")
plt.xlabel("X1")
plt.ylabel("X2")
plt.show()
```

```python
X.shape
```

```python
y.shape
```

```python
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X)
```

```python
plt.figure(figsize=(6,6))
plt.scatter(X_pca[:,0], X_pca[:,1], c=y, cmap='viridis')
plt.title("After PCA")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```

```python
kpca = KernelPCA(
    n_components=2,
    kernel='rbf',
    gamma=15
)

X_kpca = kpca.fit_transform(X)
```

```python
plt.figure(figsize=(6,6))
plt.scatter(X_kpca[:,0], X_kpca[:,1], c=y, cmap='viridis')
plt.title("After Kernel PCA")
plt.xlabel("Kernel PC1")
plt.ylabel("Kernel PC2")
plt.show()
```
