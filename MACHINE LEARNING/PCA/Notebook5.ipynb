```python
from sklearn.datasets import make_swiss_roll
from sklearn.decomposition import KernelPCA
import matplotlib.pyplot as plt
```

```python
X, color = make_swiss_roll(
    n_samples=1000,
    noise=0.2,
    random_state=42
)
```

```python
X.shape
```

```python
color.shape
```

```python
# Create a 3D plot
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

scatter = ax.scatter(
    X[:, 0],    # X-axis
    X[:, 1],    # Y-axis
    X[:, 2],    # Z-axis
    c=color,
    cmap='viridis',
    s=15
)
```

```python
# Apply Kernel PCA
kpca = KernelPCA(
    n_components=2,
    kernel='rbf',
    gamma=0.01
)
```

```python
X_kpca = kpca.fit_transform(X)
```

```python
# Plot transformed data
plt.figure(figsize=(7,5))
plt.scatter(X_kpca[:,0], X_kpca[:,1], c=color, cmap='viridis')
plt.title("Swiss Roll after Kernel PCA")
plt.xlabel("Kernel PC1")
plt.ylabel("Kernel PC2")
plt.show()
```
