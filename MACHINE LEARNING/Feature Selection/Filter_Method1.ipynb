1. Variance Threshold

```python
from sklearn.feature_selection import VarianceThreshold

selector= VarianceThreshold(threshold= 0.1)
X_new = selector.fit_transform(X)

print("New :",X_new.shape)
print("Old :" ,X.shape)
```

```python
X.head()
```

```python
selected = X.columns[selector.get_support()]

print(selected)
```

```python
X_new = pd.DataFrame(X_new, columns=selected)

X_new.head()
```

2- Correlation Method

```python
data = load_breast_cancer()
```

```python
print(data.keys())
```

```python
X= pd.DataFrame(data.data, columns= data.feature_names)

y= data.target
```

```python
df = X.copy()

df["target"] = y
```

```python
df.head()
```

```python
print(df.shape)
```

```python
corr_matrix = df.corr()
print(corr_matrix)
```

```python
target_corr= corr_matrix['target']
```

```python
target_corr
```

```python
target_corr = target_corr.sort_values(
    ascending=False
)

print(target_corr)
```

```python
target_corr = target_corr.drop("target")
```

```python
selected_features= target_corr[abs(target_corr)>0.5]
```

```python
print(selected_features)
```

```python
selected_columns = selected_features.index

print(selected_columns)
```

```python
X_selected = X[selected_columns]
```

```python
X_selected.head()
```

3. Chi-Square

```python
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import chi2
```

```python
data = load_breast_cancer()
```

```python
X = pd.DataFrame(
    data.data,
    columns=data.feature_names
)

y = data.target
```

```python
print(X.shape)
```

```python
selector = SelectKBest(
    score_func=chi2,
    k=10
)

# Selecting 10 best features
```

```python
X_new = selector.fit_transform(X, y)
```

```python
print(X_new.shape)
```

```python
selected_columns = X.columns[
    selector.get_support()
]

print(selected_columns)
```

```python
X_new = pd.DataFrame(
    X_new,
    columns=selected_columns
)
```

```python
scores = selector.scores_

print(scores)
```

```python
feature_scores = pd.DataFrame({
    "Feature": X.columns,
    "Chi2 Score": selector.scores_
})

print(feature_scores)
```

```python
feature_scores = feature_scores.sort_values(
    by="Chi2 Score",
    ascending=False
)

print(feature_scores)
```
