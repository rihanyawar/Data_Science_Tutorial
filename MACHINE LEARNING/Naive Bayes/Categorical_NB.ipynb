```python
import pandas as pd

from sklearn.preprocessing import OrdinalEncoder
from sklearn.naive_bayes import CategoricalNB

# Dataset
df = pd.DataFrame({
    "Weather": ["Sunny","Sunny","Rainy","Cloudy","Rainy","Sunny"],
    "Windy": ["Yes","No","No","Yes","Yes","No"],
    "Play": ["No","Yes","Yes","Yes","No","Yes"]
})
```

```python
X = df[["Weather","Windy"]]
y = df["Play"]
```

```python
# Convert categories to integers
encoder = OrdinalEncoder()
X_encoded = encoder.fit_transform(X)

print(X_encoded)
```

```python
# Train
model = CategoricalNB()

model.fit(X_encoded, y)

# New sample
new = pd.DataFrame({
    "Weather":["Sunny"],
    "Windy":["No"]
})

new_encoded = encoder.transform(new)

prediction = model.predict(new_encoded)

print(prediction)
```
