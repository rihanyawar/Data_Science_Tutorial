```python
#Feature engineering
#Separating the date column values because the model will not understand it
df['Date']=df['Date_of_Journey'].str.split('/').str[0]
df['Month']=df['Date_of_Journey'].str.split('/').str[1]
df['Year']=df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.head()
```

```python
df.info()
```

```python
df['Date']=df['Date'].astype(int)
df['Month']=df['Month'].astype(int)
df['Year']=df['Year'].astype(int)
```

```python
df.info()
```

```python
df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x: x.split(' ')[0])
```

```python
df['Arrival_Time']
```

```python
df['Arrival_hour']=df['Arrival_Time'].str.split(":").str[0]
df['Arrival_min']=df['Arrival_Time'].str.split(":").str[1]
```

```python
df['Arrival_hour']=df['Arrival_hour'].astype(int)
df['Arrival_min']=df['Arrival_min'].astype(int)
```

```python
df.drop("Arrival_Time",axis=1,inplace=True)
df
```

```python
#Now same for departure time
df['Dep_hour']=df['Dep_Time'].str.split(":").str[0]
df['Dep_min']=df['Dep_Time'].str.split(":").str[1]
```

```python
df['Dep_hour']=df['Dep_hour'].astype(int)
df['Dep_min']=df['Dep_min'].astype(int)
df.drop("Dep_Time",axis=1,inplace=True)
df.head()
```

```python
df.info()
```

```python
df['Total_Stops'].unique()
```

```python
df.isnull().sum()
```

```python
df[df['Total_Stops'].isnull()]
```

```python
df['Total_Stops'].mode()
```

```python
#Replace the NaN value with one stop
df['Total_Stops']=df['Total_Stops'].map({'non-stop':0, '2 stops':2, '1 stop':1, '3 stops':3, np.nan:1, '4 stops':4})
```

```python
df[df['Total_Stops'].isnull()]
```

```python
df
```

```python
#Drop route because we have the source and destination
df.drop('Route', axis=1, inplace=True)
df.head()
```

```python
df['Duration_hour']=df['Duration'].str.split(' ').str[0].str.split('h').str[0]
df['Duration_min']=df['Duration'].str.split(' ').str[1].str.split('m').str[0]
```

```python
df.info()
```

```python
df['Duration_hour'].unique()
```

```python
df['Duration_hour'].mode()
```

```python
df['Duration_min'].mode()
```

```python
#Now replace the 5m with 5
df['Duration_hour']=df['Duration_hour'].replace('5m',5 )
df['Duration_min']=df['Duration_min'].replace(np.nan,30 )
df['Duration_min'].unique()
```

```python
df['Duration_hour']=df['Duration_hour'].astype(int)
df['Duration_min']=df['Duration_min'].astype(int)
df.drop('Duration', axis=1, inplace=True)
df.head()
```

```python
df['Airline'].unique()
```

```python
df['Source'].unique()
```

```python
df['Additional_Info'].unique()
```

Now converting all these categorical features into numerical features by using OneHot Encoding

```python
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder()
encoder.fit_transform(df[['Airline', 'Source', 'Destination', 'Additional_Info']]).toarray()
```

```python
df2=pd.DataFrame(encoder.fit_transform(df[['Airline', 'Source', 'Destination', 'Additional_Info']]).toarray(),
             columns=encoder.get_feature_names_out(['Airline', 'Source', 'Destination', 'Additional_Info']))
```

```python
df2
```

```python
df.drop(['Airline', 'Source', 'Destination', 'Additional_Info'], axis=1, inplace=True)
df
```

```python
#Now merge both data frames
df_final=pd.concat([df, df2], axis=1)
df_final.head()
```
